```python
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
```

```python
# -------------------------------------------------
# Create FrozenLake Environment
# -------------------------------------------------
env_desc = [
    "SFFF",
    "FHFH",
    "FFFH",
    "HFFG"
]

env = gym.make("FrozenLake-v1", desc=env_desc, is_slippery=True)
# write your code here
```

```python
# -------------------------------------------------
# Value Iteration Algorithm
# -------------------------------------------------

def value_iteration(env, gamma=0.99, theta=1e-8):
    """
    Performs value iteration and returns the optimal value function.
    """
    num_states = env.observation_space.n
    num_actions = env.action_space.n
    V = np.zeros(num_states)
    iteration = 0

    while True:
        delta = 0
        for s in range(num_states):
            v = V[s]
            q_values = np.zeros(num_actions)
            for a in range(num_actions):
                for prob, next_state, reward, done in env.P[s][a]:
                    q_values[a] += prob * (reward + gamma * V[next_state] * (not done))
            V[s] = np.max(q_values)
            delta = max(delta, abs(v - V[s]))

        iteration += 1
        if delta < theta:
            break

    # Derive the optimal policy based on the optimal value function
    policy = np.zeros(num_states, dtype=int)
    for s in range(num_states):
        q_values = np.zeros(num_actions)
        for a in range(num_actions):
            for prob, next_state, reward, done in env.P[s][a]:
                q_values[a] += prob * (reward + gamma * V[next_state] * (not done))
        policy[s] = np.argmax(q_values)

    return V, policy, iteration
```

```python
# -------------------------------------------------
# Run Value Iteration
# -------------------------------------------------

V, policy, iterations = value_iteration(env.unwrapped)
```
